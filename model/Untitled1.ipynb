```python
aqi_data={
    "ihi":[50,100,200,300,400,500],
    "ilo":[0,51,101,201,301,401],
    "bphi2.5":[30,60,90,120,250,550],
    "bplo2.5":[0,31,61,91,121,251],
    "bphipm10":[50,100,250,350,430,500],
    "bplopm10":[0,51,101,251,351,431],
    "bphico":[1,2,10,17,34,50],
    "bploco":[0,1.1,2.1,10.1,17.1,50.1],
    "bphio3":[50,100,168,208,748,1000],
    "bploo3":[0,51,101,169,209,749],
    "bphino2":[40,80,180,280,400,600],
    "bplono2":[0,41,81,181,281,401],
    "bphiso2":[40,80,380,800,1600,2000],
    "bploso2":[0,41,81,381,801,1601],
    "bphinh3":[200,400,800,1200,1800,2000],
    "bplonh3":[0,201,401,801,1201,1801]
}
```

```python
aqi_data=pd.DataFrame(aqi_data)
aqi_data
```

```python
def aqi25(a,aqi_data):
    row=aqi_data[(aqi_data["bplo2.5"] <= a) & (aqi_data["bphi2.5"] >= a)]
    ihi=aqi_data["ihi"].iloc[0]
    ilo=aqi_data["ilo"].iloc[0]
    bphi2_5=aqi_data["bphi2.5"].iloc[0]
    bplo2_5=aqi_data["bplo2.5"].iloc[0]
    aqi2_5=((ihi - ilo) / (bphi2_5 - bplo2_5)) * (a - bplo2_5) + ilo
    return aqi2_5
```

```python
def aqi10(a,aqi_data):
    row=aqi_data[(aqi_data["bplopm10"] <= a) & (aqi_data["bphipm10"] >= a)]
    ihi=aqi_data["ihi"].iloc[0]
    ilo=aqi_data["ilo"].iloc[0]
    bphi10=aqi_data["bphipm10"].iloc[0]
    bplo10=aqi_data["bplopm10"].iloc[0]
    aqi10=((ihi - ilo) / (bphi10 - bplo10)) * (a - bplo10) + ilo
    return aqi10
```

```python
def aqio3(a,aqi_data):
    row=aqi_data[(aqi_data["bploo3"] <= a) & (aqi_data["bphio3"] >= a)]
    ihi=aqi_data["ihi"].iloc[0]
    ilo=aqi_data["ilo"].iloc[0]
    bphio3=aqi_data["bphio3"].iloc[0]
    bploo3=aqi_data["bploo3"].iloc[0]
    aqio3=((ihi - ilo) / (bphio3 - bploo3)) * (a - bploo3) + ilo
    return aqio3
```

```python
def aqico(a,aqi_data):
    row=aqi_data[(aqi_data["bploco"] <= a) & (aqi_data["bphico"] >= a)]
    ihi=aqi_data["ihi"].iloc[0]
    ilo=aqi_data["ilo"].iloc[0]
    bphico=aqi_data["bphico"].iloc[0]
    bploco=aqi_data["bploco"].iloc[0]
    aqico=((ihi - ilo) / (bphico - bploco)) * (a - bploco) + ilo
    return aqico
```

```python
def aqino2(a,aqi_data):
    row=aqi_data[(aqi_data["bplono2"] <= a) & (aqi_data["bphino2"] >= a)]
    ihi=aqi_data["ihi"].iloc[0]
    ilo=aqi_data["ilo"].iloc[0]
    bphino2=aqi_data["bphino2"].iloc[0]
    bplono2=aqi_data["bplono2"].iloc[0]
    aqino2=((ihi - ilo) / (bphino2 - bplono2)) * (a - bplono2) + ilo
    return aqino2
```

```python
def aqiso2(a,aqi_data):
    row=aqi_data[(aqi_data["bploso2"] <= a) & (aqi_data["bphiso2"] >= a)]
    ihi=aqi_data["ihi"].iloc[0]
    ilo=aqi_data["ilo"].iloc[0]
    bphiso2=aqi_data["bphiso2"].iloc[0]
    bploso2=aqi_data["bploso2"].iloc[0]
    aqiso2=((ihi - ilo) / (bphiso2 - bploso2)) * (a - bploso2) + ilo
    return aqiso2
```

```python
def aqinh3(a,aqi_data):
    row=aqi_data[(aqi_data["bplonh3"] <= a) & (aqi_data["bphinh3"] >= a)]
    ihi=aqi_data["ihi"].iloc[0]
    ilo=aqi_data["ilo"].iloc[0]
    bphinh3=aqi_data["bphinh3"].iloc[0]
    bplonh3=aqi_data["bplonh3"].iloc[0]
    aqinh3=((ihi - ilo) / (bphinh3 - bplonh3)) * (a - bplonh3) + ilo
    return aqinh3
```

```python
# spiliting for training and testing of data
X_train=X.iloc[:round(len(X)*0.8)]
X_test=X.iloc[round(len(X)*0.8):]
```

```python
y25_train=y_25.iloc[:round(len(X)*0.8)]
y25_test=y_25.iloc[round(len(X)*0.8):]
```

```python
y10_train=y_10.iloc[:round(len(X)*0.8)]
y10_test=y_10.iloc[round(len(X)*0.8):]
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_25 = RandomForestRegressor(
    n_estimators=300,
    max_depth=12,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)
rf_25.fit(X_train,y25_train)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_10 = RandomForestRegressor(
    n_estimators=300,
    max_depth=12,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)
rf_10.fit(X_train,y10_train)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
y25_pred=rf_25.predict(X_test)
y10_pred=rf_10.predict(X_test)
```

```python
k = X.shape[1]
n = len(y25_test)
mse25=mean_squared_error(y25_test,y25_pred)
mae25=mean_absolute_error(y25_test,y25_pred)
r2_25=r2_score(y25_test,y25_pred)
adj_r2_25=1-(((1-r2_25)*(n-1))/(n-k-1))
mse10=mean_squared_error(y10_test,y10_pred)
mae10=mean_absolute_error(y10_test,y10_pred)
r2_10=r2_score(y10_test,y10_pred)
adj_r2_10=1-(((1-r2_10)*(n-1))/(n-k-1))
```

```python
print(f"mse25={mse25},mae25={mae25},r2_25={r2_25},adj_r225={adj_r2_25}")
print(f"mse10={mse10},mae10={mae10},r2_10={r2_10},adj_r210={adj_r2_10}")
```
